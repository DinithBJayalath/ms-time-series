```python
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd
```

```python
energy = pd.read_csv("data/energy.csv")
energy.drop(["temp"], inplace=True, axis=1)
energy.set_index("timestamp", inplace=True)
energy.head()
```

```python
energy.plot(y="load", subplots=True, figsize=(12,6), fontsize=12)
plt.xlabel("timestamp", fontsize=12)
plt.ylabel("load", fontsize=12)
plt.show()
```

```python
energy['2014-07-01' : '2014-07-07'].plot(y="load", subplots=True, figsize=(12, 6), fontsize=12)
plt.xlabel("timestamp", fontsize=12)
plt.ylabel("load", fontsize=12)
plt.show()
```
